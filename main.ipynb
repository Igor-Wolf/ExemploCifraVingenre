```python
def vigenre_cypher(text, key):
    encrypted_text = ""
    key_index = 0 
    key = key.upper() #Garantir que a chave seja sempre maúscula

    for char in text:
        if char.isalpha(): #Verifica se o char é uma letra
            shift = ord(key[key_index]) - ord('A')
            if char.isupper():
                new_char = chr((ord(char) - ord('A') + shift) %26 + ord('A'))
            else :
                new_char = chr((ord(char) - ord('a') + shift) %26 + ord('a'))
            encrypted_text += new_char
            key_index = (key_index + 1) % len(key)
        else :
            encrypted_text += char

    
    return encrypted_text


```

```python
texto = " E ai Arthur"
chave = "MANO"
texto_cifrado = vigenre_cypher(texto, chave)
print(texto_cifrado)
```
